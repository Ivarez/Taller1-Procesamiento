### Solución del Examen sobre Operaciones con Cadenas
A continuación, se presenta la resolución de los ejercicios planteados en el taller, documentando la ejecución de cada instrucción.

```python
# Pregunta 1, 2 y 3: Asignación y concatenación
A = "1"
B = "2"
# Se concatenan las cadenas de texto, en lugar de realizar una suma aritmética.
C = A + B
print("El valor de C es:", C)
```

```python
# Pregunta 4: Slicing para imprimir los primeros tres elementos
D = "ABCDEFG"
# Se aplica el slicing indicando el inicio y el corte en la posición 3.
# Se tiene en cuenta que la indexación en Python comienza en 0.
print("Primeros tres elementos:", D[:3])
```

```python
# Pregunta 5: Uso de stride
E = 'clocrkr1e1c1t'
# Se utiliza el valor de stride [::2] para extraer la cadena saltando de dos en dos elementos.
print("Cadena con salto de 2:", E[::2])
```

```python
# Pregunta 6: Imprimir una barra invertida
# Se emplea una doble barra invertida, o el prefijo 'r', para evitar secuencias de escape.
print(r"\")
```

```python
# Pregunta 7: Conversión a mayúsculas
F = "You are wrong"
# Se utiliza el método upper() para estandarizar el texto, una operación común en bases de datos.
print("Todo en mayúsculas:", F.upper())
```

```python
# Pregunta 8 y 9: Búsqueda y reemplazo de sub-cadenas
G = "Mary had a little lamb..." # (Texto acortado para el ejemplo)
# Se busca el índice inicial de la sub-cadena 'snow'.
indice_nieve = G.find('snow')

# Se sustituyen todas las ocurrencias de 'Mary' por 'Bob' usando replace().
G_nuevo = G.replace('Mary', 'Bob')
```
