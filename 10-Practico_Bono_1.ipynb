### Calentamiento
A continuación, se documenta la resolución analítica de los ejercicios de calentamiento. Todas las operaciones están probadas y comentadas según la estructura lógica empleada.

```python
# MENOR DE DOS PARES: Se evalúa la paridad de ambos argumentos numéricos mediante el operador módulo.
def lesser_of_two_evens(a, b):
    # Si ambos son pares (módulo 2 igual a cero), se retorna el menor.
    if a % 2 == 0 and b % 2 == 0:
        return min(a, b)
    # En caso de que al menos uno sea impar, se retorna el mayor.
    else:
        return max(a, b)

# Bloque de revisión
print("lesser_of_two_evens(2, 4) ->", lesser_of_two_evens(2, 4))
print("lesser_of_two_evens(2, 5) ->", lesser_of_two_evens(2, 5))
```

```python
# GALLETAS DE ANIMALES: Se separa la cadena en una lista de palabras y se compara el primer índice de ambas en formato minúscula.
def animal_crackers(text):
    palabras = text.split()
    # Se valida que existan al menos dos elementos para evitar errores de índice.
    if len(palabras) >= 2:
        return palabras[0][0].lower() == palabras[1][0].lower()
    return False

# Bloque de revisión
print("animal_crackers('Ciudad Calor') ->", animal_crackers('Ciudad Calor'))
print("animal_crackers('Lugar Frio') ->", animal_crackers('Lugar Frio'))
```

```python
# HACE VEINTE: Se emplea una evaluación lógica con el operador 'or' para verificar las precondiciones estipuladas.
def makes_twenty(n1, n2):
    return n1 == 20 or n2 == 20 or (n1 + n2) == 20

# Bloque de revisión
print("makes_twenty(20, 10) ->", makes_twenty(20, 10))
print("makes_twenty(2, 3) ->", makes_twenty(2, 3))
print("makes_twenty(12, 8) ->", makes_twenty(12, 8))
```

### Nivel 1
Ejercicios enfocados en la manipulación y transformación nativa de cadenas de texto (Strings).

```python
# MAYÚSCULAS: Se aplica slicing para subdividir la cadena y se utiliza el método capitalize() sobre cada sección antes de concatenarlas.
def old_macdonald(name):
    # Se asegura que la longitud sea suficiente para tener un cuarto carácter (índice 3).
    if len(name) > 3:
        return name[:3].capitalize() + name[3:].capitalize()
    return name.capitalize()

# Bloque de revisión
print("old_macdonald('macdonald') ->", old_macdonald('macdonald'))
```

```python
# REVERSA: Se extraen las palabras en una lista, se invierte el orden mediante slicing negativo y se reensambla con join().
def master_yoda(text):
    palabras = text.split()
    return " ".join(palabras[::-1])

# Bloque de revisión
print("master_yoda('I am home') ->", master_yoda('I am home'))
print("master_yoda('We are ready') ->", master_yoda('We are ready'))
```

### Nivel 2
Ejercicios lógicos de mayor complejidad que requieren el uso de iteradores, contadores y validaciones numéricas complejas.

```python
# PROBLEMA 33: Se iteran los índices de la lista, comprobando simultáneamente el elemento actual y el adyacente.
def has_33(nums):
    # Se recorta el rango en -1 para prevenir un error de 'índice fuera de rango' en la última iteración.
    for i in range(len(nums) - 1):
        if nums[i] == 3 and nums[i + 1] == 3:
            return True
    return False

# Bloque de revisión
print("has_33([1, 3, 3]) ->", has_33([1, 3, 3]))
print("has_33([1, 3, 1, 3]) ->", has_33([1, 3, 1, 3]))
print("has_33([3, 1, 3]) ->", has_33([3, 1, 3]))
```

```python
# REPLICADOR: Se recorre la cadena de texto, multiplicando cada caracter individual para construir la nueva salida estructurada.
def paper_doll(text):
    resultado = ""
    for char in text:
        resultado += char * 3
    return resultado

# Bloque de revisión
print("paper_doll('Hello') ->", paper_doll('Hello'))
print("paper_doll('Mississippi') ->", paper_doll('Mississippi'))
```

```python
# BLACKJACK: Se agrupan los enteros, calculando sumatorias y evaluando el flujo mediante múltiples ramificaciones if-elif-else.
def blackjack(a, b, c):
    suma = sum((a, b, c))
    
    if suma <= 21:
        return suma
    # Se ajusta la reducción si la suma excede 21 y existe un '11' en los operandos iniciales.
    elif suma > 21 and 11 in (a, b, c):
        return suma - 10
    else:
        return 'BUST'

# Bloque de revisión
print("blackjack(5, 6, 7) ->", blackjack(5, 6, 7))
print("blackjack(9, 9, 9) ->", blackjack(9, 9, 9))
print("blackjack(9, 9, 11) ->", blackjack(9, 9, 11))
```
