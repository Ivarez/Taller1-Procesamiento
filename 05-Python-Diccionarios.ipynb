En este quinto cuaderno se detalla el uso de diccionarios. Estas estructuras operan mediante un mapeo de llaves y valores en lugar de índices numéricos. Son esenciales cuando se requiere etiquetar y acceder a información específica mediante identificadores legibles, asemejándose a los registros de una base de datos.

### Solución del Examen sobre Diccionarios en Python
A continuación, se presenta la resolución de los ejercicios planteados en el taller, documentando la ejecución de cada instrucción.

```python
# Pregunta 1a y 1b: Identificar llaves y valores
soundtrack_dic = {"The Bodyguard":"1992", "Saturday Night Fever":"1977"}
# Se utilizan los métodos nativos para extraer ambas secuencias del diccionario.
llaves_soundtrack = soundtrack_dic.keys()
valores_soundtrack = soundtrack_dic.values()
print("Llaves:", llaves_soundtrack)
print("Valores:", valores_soundtrack)
```

```python
# Pregunta 2a: Crear el diccionario album_sales_dict
# Se asignan las llaves (cadenas de texto) a sus respectivos valores (números enteros).
album_sales_dict = {"Back in Black": 50, "The Bodyguard": 50, "Thriller": 65}
print("Diccionario de ventas:", album_sales_dict)
```

```python
# Pregunta 2b: Encontrar las ventas totales de Thriller
# Se accede al valor utilizando la llave exacta dentro de corchetes.
ventas_thriller = album_sales_dict["Thriller"]
print("Ventas de Thriller (millones):", ventas_thriller)
```

```python
# Pregunta 2c y 2d: Encontrar nombres y récords mediante métodos
# Se extraen independientemente las listas de llaves y valores.
nombres_discos = album_sales_dict.keys()
records_ventas = album_sales_dict.values()
print("Nombres de los discos:", nombres_discos)
print("Récords de ventas:", records_ventas)
```

### Conclusiones
Los diccionarios me resultan indispensables para aquellos escenarios donde la indexación secuencial no es suficiente. El uso de llaves inmutables para mapear valores me permite estructurar la información de manera sumamente organizada y explícita. Reconozco que utilizar funciones como `keys()` y `values()` agiliza mucho la recuperación de los datos, garantizando consultas eficientes que serán clave al momento de integrar grandes volúmenes de registros.
