En este séptimo cuaderno se aborda la iteración a través de bucles `for` y `while`. La automatización de tareas repetitivas mediante ciclos es estrictamente necesaria para recorrer colecciones de datos masivas y aplicar transformaciones sistemáticas a gran escala.

### Solución del Examen sobre Bucles en Python
A continuación, se documenta la resolución analítica de los ejercicios planteados en el taller.

```python
# Pregunta 1: Bucle for con rango
# Se utiliza la función range() estableciendo el límite superior en 6, ya que este es exclusivo.
for i in range(-5, 6):
    print(i)
```

```python
# Pregunta 2: Iteración sobre una lista de cadenas
Genres = ['rock', 'R&B', 'Soundtrack', 'R&B', 'soul', 'pop']
# Se recorre la lista asignando secuencialmente cada elemento a la variable temporal 'genero'.
for genero in Genres:
    print(genero)
```

```python
# Pregunta 3: Iteración e impresión
squares = ['red', 'yellow', 'green', 'purple', 'blue']
# Se emplea el mismo paradigma iterativo para mostrar el contenido de las variables.
for square in squares:
    print(square)
```

```python
# Pregunta 4: Bucle while con condición de corte numérica
PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]
i = 0
# El bucle se mantiene en ejecución mientras no se desborde el índice y la calificación sea mayor o igual a 6.
while i < len(PlayListRatings) and PlayListRatings[i] >= 6:
    print(PlayListRatings[i])
    i += 1
```

```python
# Pregunta 5: Bucle while con condición de igualdad de cadena
squares_2 = ['orange', 'orange', 'purple', 'blue ', 'orange']
new_squares = []
i = 0
# Se clonan los elementos a la nueva lista hasta que la evaluación lógica detecte una cadena diferente.
while i < len(squares_2) and squares_2[i] == 'orange':
    new_squares.append(squares_2[i])
    i += 1
print("Elementos copiados:", new_squares)
```

### Conclusiones
Comprender la diferencia operativa entre los bucles `for` y `while` considero que es vital para la estructuración de scripts eficientes. Mientras que el ciclo `for` me resulta ideal para recorrer colecciones de longitud conocida o iterables definidos, el ciclo `while` ofrece una flexibilidad analítica superior cuando la interrupción de la iteración depende estrictamente de cambios de estado evaluados en tiempo real.
