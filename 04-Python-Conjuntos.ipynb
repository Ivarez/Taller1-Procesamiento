En este cuarto cuaderno se revisa el concepto de los conjuntos (sets). Estas colecciones se caracterizan por almacenar objetos únicos, eliminando automáticamente cualquier duplicado. Resultan extremadamente útiles para filtrar datos y realizar operaciones lógicas de teoría de conjuntos.

### Solución del Examen sobre Conjuntos en Python
A continuación, se presenta la resolución de los ejercicios planteados en el taller, documentando la ejecución de cada instrucción.

```python
# Pregunta 1: Transformar una lista en un conjunto
lista_musica = ['rap', 'house', 'electronic music', 'rap']
# Se utiliza la función set() para convertir la lista, eliminando el elemento duplicado ('rap').
conjunto_musica = set(lista_musica)
print("Conjunto generado:", conjunto_musica)
```

```python
# Pregunta 2: Comparar la suma de una lista y un conjunto
A = [1, 2, 2, 1]
B = set([1, 2, 2, 1])
# Se comprueba que la suma difiere, ya que el conjunto solo conserva valores únicos (1 y 2).
print("¿Las sumas son iguales?:", sum(A) == sum(B))
```

```python
# Pregunta 3: Crear la unión de dos conjuntos
album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set(["AC/DC", "Back in Black", "The Dark Side of the Moon"])
# Se emplea el método union() para consolidar todos los elementos sin repeticiones.
album_set3 = album_set1.union(album_set2)
print("Unión de conjuntos (album_set3):", album_set3)
```

```python
# Pregunta 4: Averiguar si album_set1 es subconjunto de album_set3
# Se aplica issubset() para verificar la jerarquía entre las colecciones.
es_subconjunto = album_set1.issubset(album_set3)
print("¿album_set1 es subconjunto de album_set3?:", es_subconjunto)
```
