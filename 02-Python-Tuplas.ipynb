En este segundo cuaderno se explora el uso de las tuplas. Estas estructuras son similares a las listas, pero se distinguen por ser inmutables. Esta característica las hace ideales para almacenar información crítica que no debe ser alterada durante la ejecución del programa.

### Solución del Examen sobre Tuplas en Python
A continuación, se presenta la resolución de los ejercicios planteados en el taller, documentando la ejecución de cada instrucción.

```python
# Se define la tupla de ejemplo
genres_tuple = ("pop", "rock", "soul", "hard rock", "soft rock", "R&B", "progressive rock", "disco")
```

```python
# Pregunta 1: Encontrar la longitud
# Se emplea la función len() para determinar la cantidad exacta de elementos.
longitud = len(genres_tuple)
```

```python
# Pregunta 2: Acceder al elemento en el índice 3
# Se extrae el cuarto elemento real, recordando la indexación desde cero.
elemento_3 = genres_tuple[3]
```

```python
# Pregunta 3: Slicing para obtener los índices 3, 4 y 5
# Se define el rango [3:6] dado que el límite superior no es inclusivo.
pedazo_tupla = genres_tuple[3:6]
```

```python
# Pregunta 4: Encontrar los primeros dos elementos
# Se realiza el corte desde el inicio hasta la posición 2 (exclusiva).
primeros_dos = genres_tuple[0:2]
```

```python
# Pregunta 5: Encontrar el primer índice de "disco"
# Se utiliza el método index() para localizar la posición de la cadena especificada.
indice_disco = genres_tuple.index("disco")
```

```python
# Pregunta 6: Generar una lista ordenada desde una tupla
C_tuple = (-5, 1, -3)
# Dado que la tupla es inmutable, se aplica sorted() para generar una nueva estructura tipo lista.
lista_ordenada = sorted(C_tuple)
```

### Conclusiones
Considero que las tuplas son una de las estructuras más seguras dentro del lenguaje debido a su inmutabilidad. Aunque en un principio no poder reasignar valores puede parecer restrictivo, aplicar conceptos como la indexación y el slicing me permite acceder y subdividir la información sin comprometer la integridad de los datos originales. El uso de funciones como `sorted()` también me demuestra que siempre es posible transformar y operar con estos datos cuando se necesita flexibilidad.
