### Solución del Examen sobre Clases y Objetos en Python
A continuación, se documenta la resolución analítica de los ejercicios planteados en el taller.

```python
# Ejercicio Único: Creación de la clase Elipse y sus métodos
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

class Elipse(object):
    # Constructor con asignación de atributos iniciales predeterminados
    def __init__(self, width=2, height=1, facecolor='blue', edgecolor='black'):
        self.width = width
        self.height = height
        self.facecolor = facecolor
        self.edgecolor = edgecolor

    # Métodos mutadores para modificar los atributos del objeto en memoria
    def cambiar_ancho(self, nuevo_ancho):
        self.width = nuevo_ancho

    def cambiar_alto(self, nuevo_alto):
        self.height = nuevo_alto

    def cambiar_color_relleno(self, nuevo_color):
        self.facecolor = nuevo_color

    def cambiar_color_borde(self, nuevo_borde):
        self.edgecolor = nuevo_borde

    # Método de graficación invocando la librería matplotlib
    def dibujar(self):
        # Se genera el mapeo del parche (patch) en los ejes coordenados
        plt.gca().add_patch(Ellipse((0, 0), self.width, self.height, facecolor=self.facecolor, edgecolor=self.edgecolor))
        plt.axis('scaled')
        plt.show()

# Bloque de prueba de ejecución e instanciación
mi_elipse = Elipse()
print("Elipse original:")
mi_elipse.dibujar()

# Aplicación de los métodos creados
mi_elipse.cambiar_ancho(4)
mi_elipse.cambiar_color_relleno('green')
mi_elipse.cambiar_color_borde('red')
print("Elipse modificada:")
mi_elipse.dibujar()
```
